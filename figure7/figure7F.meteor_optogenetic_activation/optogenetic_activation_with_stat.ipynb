```python
import pandas as pd 
import sys
from pathlib import Path
REPO_ROOT = Path.cwd().resolve().parents[1]
sys.path.insert(0, str(REPO_ROOT / "code"))
import plot_of_experiment_data 
import matplotlib.pyplot as plt
import matplotlib
import warnings
import numpy as np 
warnings.filterwarnings("ignore", message="findfont: Font family 'Arial' not found.")
matplotlib.rcParams['svg.fonttype'] = 'none'
matplotlib.rcParams['font.family'] = 'Arial'
matplotlib.rcParams['font.size'] = 7
```

```python

fpath = 'data/optogenetics_geosmin_pathway_251119.xlsx'
df = pd.read_excel(fpath,sheet_name='meteor SS35420')
data,pivot,error = plot_of_experiment_data.make_pivot_optogenetics_data(df)
```

```python
# target_sheets = ['meteor SS35420', 'meteor SS35412']
```

```python
fig, ax = plt.subplots(1, 1, figsize=(1.07, 1))

ax,stat_all = plot_of_experiment_data.draw_paired_bar_figure(
    data=data,
    mean_df=pivot,
    err_df=error,
    bar_width=0.1,
    bar_spacing=.3,
    stat_fontsize=7,
    linewidth=.5,
    fontsize=7,
    dot_size=10,
    ax=ax,
    conds=[0, 1],
    groups=['retinal (-)', 'retinal (+)'],
    get_values=lambda data, cond, group: data[(cond, 0 if group == 'retinal (-)' else 1)],
    stat_pairs=[
        ((0, 'retinal (+)'), (1, 'retinal (+)'))
    ],
    colors=('#AEACAB', 'black'),
    ylabel='Tarsal PER',
#     xtick_func=lambda c: '-' if c == 0 else '+',
    ylim=(0, 1),
    yticks=np.round(np.arange(0, 1.2, .2), 2)
)

xlabel = conds = ['+\n-','+\n+']
ax.set_xticklabels(xlabel)
bar_spacing = .3
num_bars = len(conds)
x_loc = np.arange(0, num_bars * bar_spacing, bar_spacing)

ax.set_xlim(-0.2, (num_bars - 1) * bar_spacing + 0.2)
ax.tick_params(
    axis='both',
    width=0.7
)

ax.spines['left'].set_position(('outward', 5))
ax.tick_params(
    axis='both',
    width=0.7
)

# Axis (spine) thickness
for spine in ax.spines.values():
    spine.set_linewidth(0.7)
    
     
    
# plt.savefig('figures/meteor_optogenetics.svg')
```
