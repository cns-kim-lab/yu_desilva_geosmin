```python
# Valid Statistical test
import pandas as pd
from scipy.stats import mannwhitneyu

results = []

for g in tarsal_s_fixed_df.columns:
    stat, p = mannwhitneyu(
        tarsal_s_fixed_df[g],
        tarsal_sg_fixed_df[g],
        alternative="two-sided",
        method="auto"
    )

    results.append({
        "group": 'movable vs. immobilized ' +g,
        "U": stat,
        "p_raw": p
    })

results = pd.DataFrame(results)

results
```

```python
# Leg immobilization shifted the tarsal sucrose-PER curve upward 
import pandas as pd
from scipy.stats import mannwhitneyu

results = []

for g in tarsal_s_fixed_df.columns:
    stat, p = mannwhitneyu(
        tarsal_s_df[g],
        tarsal_s_fixed_df[g],
        alternative="less",
        method="auto"
    )

    results.append({
        "group": 'Suc. movable vs. immobilized ' +g,
        "U": stat,
        "p_raw": p
    })

results = pd.DataFrame(results)

results
```

```python
# Leg immobilization attenuated geosmin-dependent suppression
delta_movable = tarsal_s_df - tarsal_sg_df
delta_immobilized = tarsal_s_fixed_df - tarsal_sg_fixed_df

stat, p = mannwhitneyu(
    delta_movable,
    delta_immobilized,
    alternative='greater'
)
stat,p
```

```python
# Meteor silencing effect 
delta_movable = tarsal_s_df - tarsal_sg_df
delta_meteor_slnc = tarsal_s_slnc_df - tarsal_sg_slnc_df
stat, p = mannwhitneyu(
    delta_movable,
    delta_meteor_slnc,
    alternative='greater'
)
stat,p
```

```python
# Phenocopying 

delta_meteor = tarsal_s_slnc_df- tarsal_sg_slnc_df
delta_immobilized_meteor = tarsal_s_fixed_slnc_df - tarsal_sg_fixed_slnc_df

stat, p = mannwhitneyu(
    delta_meteor,
    delta_immobilized_meteor,
    alternative='two-sided'
)
stat,p
```
