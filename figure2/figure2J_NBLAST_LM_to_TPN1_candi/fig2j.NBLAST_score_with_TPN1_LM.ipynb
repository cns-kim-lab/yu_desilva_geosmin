```python
import pickle 
import os 
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import matplotlib
matplotlib.rcParams['svg.fonttype'] = 'none'
matplotlib.rcParams['font.family'] = 'Arial'
fontsize = 10
```

```python
score_mean_df_TPN1 = pickle.load(open(f'data/score_mean_df_TPN1_ascending.pkl','rb'))
index = list(score_mean_df_TPN1.index)
score_mean_df_TPN1.index = index
```

```python
# M 
fig,axM= plt.subplots(1,1,figsize=(1.5,1.1))
fontsize=10
score_mean_df_TPN1 = score_mean_df_TPN1.sort_values(ascending=False)
score_mean_df_TPN1.plot.bar(color=['#707070'],ax=axM)

lp=2
cmap = index#['#cfce2d','#ff00ff','#007fff','#ff7f00','#7fbf7f','#62019d','#fdfc3d']
colors = score_mean_df_TPN1.index.values
ranks = [cmap.index(x) for x in colors]
axM.set_xticklabels([f'{x+1}' for x in ranks],fontsize=fontsize)
for i,tick in enumerate(axM.get_xticklabels()):
    tick.set_color(colors[i])
yticks = list(np.round(np.arange(0,7,0.3),3))
axM.set_yticks(yticks,yticks,fontsize=fontsize)
axM.tick_params('x',rotation=0)
axM.set_ylabel('NBLAST score',fontsize=fontsize,labelpad=lp)
axM.set_xlabel('putative TPN1 groups',fontsize=fontsize,labelpad=lp)
axM.tick_params('x',rotation=0)
axM.spines['top'].set_visible(False)
axM.spines['right'].set_visible(False)
axM.set_ylim([0,.6])
# plt.savefig(f'{os.getcwd()}/250714/nblast_score_TPN1.svg',bbox_inches='tight')
```
