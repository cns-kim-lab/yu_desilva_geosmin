```python
import matplotlib.pyplot as plt
import distinctipy
import numpy as np 
import pandas as pd 
import os 
import pickle
import matplotlib
from scipy.stats import dunnett
import sys
from pathlib import Path
REPO_ROOT = Path.cwd().resolve().parents[1]
sys.path.insert(0, str(REPO_ROOT / "code"))
from plot_of_experiment_data import *
from fitting import sigmoid

matplotlib.rcParams['svg.fonttype'] = 'none'
matplotlib.rcParams['font.family'] = 'Arial'
matplotlib.rcParams['font.size'] = 7


from make_network import sez
sez_ids = np.concatenate(list(sez.values()))
sez_id2g = {f:g for g,fid in sez.items() for f in fid}
```

```python
df_tarsal_simul = pd.read_parquet('data/optogenetics_simulation_geosmin_tarsal.parquet')

fpath = 'data/tarsal.xlsx'
sheets = pd.ExcelFile(fpath)
sheets = sheets.sheet_names
data_df = {}
for s in sheets:
    df = pd.read_excel(fpath,s)
    data_df[s] = df
    
pivot_per_type = {}
for s in sheets:
    this_df = data_df[s]
    data,pivot,error = make_pivot_optogenetics_data(this_df)
    pivot_per_type[s] = pivot
    
```

```python
a = 1 - pivot_per_type['horntail SS51950'].loc[1,'retinal (+)']/pivot_per_type['horntail SS51950'].loc[0,'retinal (+)']
b = 1 - pivot_per_type['web SS41446'].loc[1,'retinal (+)']/pivot_per_type['web SS41446'].loc[0,'retinal (+)']
c = 1 - pivot_per_type['meteor SS35412'].loc[1,'retinal (+)']/pivot_per_type['meteor SS35412'].loc[0,'retinal (+)']
exps = [a,b,c]

a = 1 - df_tarsal_simul.loc['horntail','88.4Hz_82.5Hz_60Hz']/df_tarsal_simul.loc['horntail','88.4Hz_82.5Hz_0Hz']
b = 1 - df_tarsal_simul.loc['web','88.4Hz_82.5Hz_60Hz']/df_tarsal_simul.loc['web','88.4Hz_82.5Hz_0Hz']
c = 1 - df_tarsal_simul.loc['meteor','88.4Hz_82.5Hz_60Hz']/df_tarsal_simul.loc['meteor','88.4Hz_82.5Hz_0Hz']
simul = [a,b,c]
```

```python
import matplotlib.pyplot as plt
import matplotlib
matplotlib.rcParams['svg.fonttype'] = 'none'
matplotlib.rcParams['font.family'] = 'Arial'
matplotlib.rcParams['font.size'] = 7
fontsize = 7 


LEFT = 0.22
RIGHT = 0.98
TOP = 0.95
BOTTOM = 0.22
fig,ax = plt.subplots(1,1,figsize=(.9,.9))
ax.scatter(simul,exps,color='#ff0088',s=15,lw=.5,edgecolors='black')
ax.plot(list(np.arange(0,1,.1)),list(np.arange(0,1,.1)),linestyle='dashed',color='red',lw=.5)
ax.set_xticks(list(np.arange(0,1.1,.5)))
ax.set_yticks(list(np.arange(0,1.1,.5)))
ax.set_xlim([0,1])
ax.set_ylim([0,1])
ax.spines[['top','right']].set_visible(False)
ax.spines['left'].set_position(('outward', 5))
ax.spines['bottom'].set_position(('outward', 5))
ax.set_ylabel('Exp.')
ax.set_xlabel('Simul.')
ax.xaxis.labelpad = 0.5
ax.yaxis.labelpad = 0.5


fig.subplots_adjust(
    left=LEFT,
    right=RIGHT,
    top=TOP,
    bottom=BOTTOM
)
# plt.savefig('optogenetic_inhibitory_neurons_tarsal.svg')
```
