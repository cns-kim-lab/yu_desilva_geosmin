```python
from pathlib import Path
import sys

REPO_ROOT = Path.cwd().resolve().parents[3]
sys.path.insert(0, str(REPO_ROOT / "code"))

from model_unified_ver import run_exp
from model_unified_ver import default_params as params
import utils as utl
from brian2 import Hz,mV
import numpy as np 
import pandas as pd 
import os 
import pickle


labial_cluster = pd.read_parquet(
    REPO_ROOT / "data" / "labial_cluster_info_v783.parquet"
)
```

```python
labial_c2g = {}
for t in list(dict.fromkeys(labial_cluster.type)):
    labial_c2g[t] = list(labial_cluster.flyid.values[labial_cluster.type==t])
labial_c2g
```

```python
for g in ['water','high_salt','L1','L2','L3']:
    params['w_syn'] = 0.275*mV
    params['n_run'] = 100
    config = {
        'path_res'  : f'{os.getcwd()}/labial_result/{g}',                              # directory to store results
        'path_comp' : REPO_ROOT / "data" /'Completeness_783.csv',        # csv of the complete list of Flywire neurons
        'path_con'  : REPO_ROOT / "data" / 'Connectivity_783.parquet',    # connectivity data
        'n_proc'    : -1,                                               # number of CPU cores (-1: use all)
    }


    if g not in os.listdir(f'{os.getcwd()}/labial_result'):
        os.mkdir(f"{config['path_res']}")
    if 'params' not in os.listdir(f'{config["path_res"]}'):
        os.mkdir(f'{config["path_res"]}/params')

    pickle.dump(params,open(f'{config["path_res"]}/params/params.pkl','wb'))


    neu_exc = labial_c2g[g]

    neu_add = []
    except_ids = None
    except_wsyn = None
    neu_slnc = []


    for f1 in list(range(0,220,20)):
        params['r_poi1'] = f1 * Hz
        run_exp(exp_name=f'{f1}Hz', neu_exc=neu_exc,neu_exc_add=neu_add,Except_output_Ids=except_ids,Except_output_w_syn=except_wsyn,neu_slnc=neu_slnc,params=params, **config)
```
