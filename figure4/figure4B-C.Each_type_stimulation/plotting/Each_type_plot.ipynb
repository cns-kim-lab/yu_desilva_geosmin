```python
import sys 
import os 
from pathlib import Path
REPO_ROOT = Path.cwd().resolve().parents[2]
sys.path.insert(0, str(REPO_ROOT / "code"))

from analysis_of_simulation_result import * 
from color_utils import make_color_map
import matplotlib.pyplot as plt
import matplotlib
from simulation_plot_utils import (
    create_candidate_stimulation_plot,
    save_figure,
)
matplotlib.rcParams['svg.fonttype'] = 'none'
matplotlib.rcParams['font.family'] = 'Arial'
matplotlib.rcParams['font.size'] = 10

atGRN_cluster = pd.read_parquet('/volume_4/research/seongbong/flywire/geosmin_project_version_update/data/atGRN_cluster_info_v783.parquet')
labial_cluster = pd.read_parquet('/volume_4/research/seongbong/flywire/geosmin_project_version_update/data/labial_cluster_info_v783.parquet')
```

```python
DATA_DIR = Path("data")
FIGURE_DIR = Path("figures")


df_mn9 = pd.read_excel(
    DATA_DIR / "Labellar_candidate_stimulation.xlsx",
    index_col="group",
)

sem_mn9 = pd.read_excel(
    DATA_DIR / "Labellar_candidate_stimulation_SEM.xlsx",
    index_col="group",
)


color_map = (
    labial_cluster[["type", "color"]]
    .drop_duplicates("type")
    .set_index("type")["color"]
    .to_dict()
)


fig, ax = create_candidate_stimulation_plot(
    mean_df=df_mn9,
    sem_df=sem_mn9,
    colors=color_map,
    figsize=(2, 2),
    xlim=(0, 200),
    ylim=(0, 80),
    xticks=range(0, 201, 40),
    spine_width=1.2,
)

save_figure(
    fig,
    FIGURE_DIR / "labellar_candidate_stimulation.svg",
)
```
