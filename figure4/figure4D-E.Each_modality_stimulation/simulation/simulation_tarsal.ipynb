```python
from pathlib import Path
import sys

REPO_ROOT = Path.cwd().resolve().parents[2]
sys.path.insert(0, str(REPO_ROOT / "code"))

from model_unified_ver import run_exp
from model_unified_ver import default_params as params
from cell_groups import *
import utils as utl
from brian2 import Hz,mV
import numpy as np 
import pandas as pd 
import os 
import pickle
 

atGRN_cluster = pd.read_parquet(
    REPO_ROOT / "data" / "atGRN_cluster_info_v783.parquet"
)

```

```python
atgrn_c2g = {}
for t in np.unique(atGRN_cluster.type):
    atgrn_c2g[t] = list(atGRN_cluster.flyid.values[atGRN_cluster.type==t])
    
```

# atGRN+TPN1

```python
params['w_syn'] = 0.275*mV
params['n_run'] = 100
config = {
    'path_res'  : f'{os.getcwd()}/tarsal_result/atGRN+TPN1',                              # directory to store results
    'path_comp' : REPO_ROOT / "data" /'Completeness_783.csv',         # csv of the complete list of Flywire neurons
    'path_con'  : REPO_ROOT / "data" / 'Connectivity_783.parquet',   # connectivity data
    'n_proc'    : -1,                                               # number of CPU cores (-1: use all)
}



if 'atGRN+TPN1' not in os.listdir(f'{os.getcwd()}/tarsal_result'):
    os.mkdir(f"{config['path_res']}")
if 'params' not in os.listdir(f'{config["path_res"]}'):
    os.mkdir(f'{config["path_res"]}/params')

pickle.dump(params,open(f'{config["path_res"]}/params/params.pkl','wb'))


neu_exc = atgrn_c2g['a6']+atgrn_c2g['a7']+TPN1

neu_add = [av1a1]
except_ids = [atgrn_c2g['a6']+atgrn_c2g['a7']+TPN1]
except_wsyn = [0.825*mV]
neu_slnc = []


for f1 in list(range(0,220,20)):
    for f2 in [0,100,150,200]:#[0,160,170,180,190]:
        params['r_poi1'] = f1 * Hz
        params['r_poi2'] = f2 * Hz
        run_exp(exp_name=f'{f1}Hz_{f2}Hz', neu_exc=neu_exc,neu_exc_add=neu_add,Except_output_Ids=except_ids,Except_output_w_syn=except_wsyn,neu_slnc=neu_slnc,params=params, **config)
```
