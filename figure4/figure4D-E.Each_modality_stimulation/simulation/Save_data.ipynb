```python
from pathlib import Path
import sys

REPO_ROOT = Path.cwd().resolve().parents[2]
sys.path.insert(0, str(REPO_ROOT / "code"))

from analysis_of_simulation_result import * 
from cell_groups import *
atGRN_cluster = pd.read_parquet(
    REPO_ROOT / "data" / "atGRN_cluster_info_v783.parquet"
)
labial_cluster = pd.read_parquet(
    REPO_ROOT / "data" / "labial_cluster_info_v783.parquet"
)


SIMULATION_DIR = Path.cwd()
OUTPUT_DIR = SIMULATION_DIR.parent / "plotting" / "data"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
N_SIMULATIONS = 100
SWEET_FREQUENCIES = list(range(0,220,20))
GEOSMIN_FREQUENCIES = [0,*list(range(100,250,50))]
result_dir = Path.cwd()/ "tarsal_result"

df_mn9, sem_mn9 = load_mn9_results(
    result_dir=result_dir,
    target_neuron=MN9,
    target_name="MN9",
    n_simulations=N_SIMULATIONS,
)

df_mn9_pivot = make_frequency_grid(
    df=df_mn9,
    sweet_frequencies=SWEET_FREQUENCIES,
    geosmin_frequencies=GEOSMIN_FREQUENCIES,
)

sem_mn9_pivot = make_frequency_grid(
    df=sem_mn9,
    sweet_frequencies=SWEET_FREQUENCIES,
    geosmin_frequencies=GEOSMIN_FREQUENCIES,
)


df_mn9_pivot.to_excel(OUTPUT_DIR/'tarsal_av1a1_simulation_result_mean.xlsx')
sem_mn9_pivot.to_excel(OUTPUT_DIR/'tarsal_av1a1_simulation_result_SEM.xlsx')




result_dir = Path.cwd() / "labial_result"

df_mn9, sem_mn9 = load_mn9_results(
    result_dir=result_dir,
    target_neuron=MN9,
    target_name="MN9",
    n_simulations=N_SIMULATIONS,
)

df_mn9_pivot = make_frequency_grid(
    df=df_mn9,
    sweet_frequencies=SWEET_FREQUENCIES,
    geosmin_frequencies=GEOSMIN_FREQUENCIES,
)

sem_mn9_pivot = make_frequency_grid(
    df=sem_mn9,
    sweet_frequencies=SWEET_FREQUENCIES,
    geosmin_frequencies=GEOSMIN_FREQUENCIES,
)



df_mn9_pivot.to_excel(OUTPUT_DIR/'labial_av1a1_simulation_result_mean.xlsx', merge_cells=False)
sem_mn9_pivot.to_excel(OUTPUT_DIR/'labial_av1a1_simulation_result_SEM.xlsx', merge_cells=False)
```
