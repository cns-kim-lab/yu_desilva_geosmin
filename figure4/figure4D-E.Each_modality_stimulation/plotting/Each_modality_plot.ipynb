```python
import sys 
import os 
from pathlib import Path
REPO_ROOT = Path.cwd().resolve().parents[2]
sys.path.insert(0, str(REPO_ROOT / "code"))

from analysis_of_simulation_result import * 
from color_utils import make_color_map
import matplotlib.pyplot as plt
import matplotlib
from simulation_plot_utils import (
    create_candidate_stimulation_plot,
    save_figure,
)
matplotlib.rcParams['svg.fonttype'] = 'none'
matplotlib.rcParams['font.family'] = 'Arial'
matplotlib.rcParams['font.size'] = 10

root_DATA_DIR = Path.cwd().resolve().parents[2]
atGRN_cluster = pd.read_parquet(root_DATA_DIR/'data/atGRN_cluster_info_v783.parquet')
labial_cluster = pd.read_parquet(root_DATA_DIR/'data/labial_cluster_info_v783.parquet')
```

```python
DATA_DIR = Path("data")
FIGURE_DIR = Path("figures")

MODALITY = "L1+L2+L3"


# ============================================================
# Load MultiIndex data
# ============================================================

mean_all = pd.read_excel(
    DATA_DIR / "labial_av1a1_simulation_result_mean.xlsx",
    index_col=[0, 1],
)

sem_all = pd.read_excel(
    DATA_DIR / "labial_av1a1_simulation_result_SEM.xlsx",
    index_col=[0, 1],
)

mean_all.index.names = ["group", "geosmin_frequency"]
sem_all.index.names = ["group", "geosmin_frequency"]


# ============================================================
# Select modality
# ============================================================

mean_df = mean_all.xs(
    MODALITY,
    level="group",
).copy()

sem_df = sem_all.xs(
    MODALITY,
    level="group",
).copy()


# ============================================================
# Labels and colors
# ============================================================

condition_label_map = {
    "0Hz": "suc.",
    "100Hz": "suc. + geo. 100 Hz",
    "150Hz": "suc. + geo. 150 Hz",
    "200Hz": "suc. + geo. 200 Hz",
}

mean_df = mean_df.rename(index=condition_label_map)
sem_df = sem_df.rename(index=condition_label_map)

condition_colors = {
    "suc.": "#26ac39",
    "suc. + geo. 100 Hz": "#f39900",
    "suc. + geo. 150 Hz": "#f39900",
    "suc. + geo. 200 Hz": "#f39900",
}

condition_alphas = {
    "suc.": 1.0,
    "suc. + geo. 100 Hz": 0.35,
    "suc. + geo. 150 Hz": 0.65,
    "suc. + geo. 200 Hz": 1.0,
}

# ============================================================
# Plot
# ============================================================

fig, ax = create_candidate_stimulation_plot(
    mean_df=mean_df,
    sem_df=sem_df,
    colors=condition_colors,
    alphas=condition_alphas,
    figsize=(2, 2),
    xlim=(0, 200),
    ylim=(0, 100),
    xticks=range(0, 201, 40),
    yticks=range(0, 101, 25),
    spine_width=1.5,
)


# ============================================================
# Save
# ============================================================


save_figure(
    fig,
    FIGURE_DIR
    / f"labellar_susceptibility_to_geosmin.svg",
)

# plt.close(fig)
```
