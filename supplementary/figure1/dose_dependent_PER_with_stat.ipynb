```python
female_s_df = pd.read_excel('data/female.xlsx',index_col=False,sheet_name='sugar')
female_sg_df = pd.read_excel('data/female.xlsx',index_col=False,sheet_name='sugar+geosmin')

male_s_df = pd.read_excel('data/male.xlsx',index_col=False,sheet_name='sugar')
male_sg_df = pd.read_excel('data/male.xlsx',index_col=False,sheet_name='sugar+geosmin')

young_s_df = pd.read_excel('data/young.xlsx',index_col=False,sheet_name='sugar')
young_sg_df = pd.read_excel('data/young.xlsx',index_col=False,sheet_name='sugar+geosmin')

old_s_df = pd.read_excel('data/old.xlsx',index_col=False,sheet_name='sugar')
old_sg_df = pd.read_excel('data/old.xlsx',index_col=False,sheet_name='sugar+geosmin')
```

```python
data_suc     = female_s_df
data_suc_geo = female_sg_df

fig,ax = plt.subplots(1,1,figsize=(1.8,2))


fig,ax = plt.subplots(1,1,figsize=(1.8,2))
plot_dose_dp_PER(ax,data_suc,**fig_params)
plot_dose_dp_PER(ax,data_suc_geo,cidx=1,**fig_params)
stat_all = plot_statistics_result(ax,[data_suc,data_suc_geo])

ax.set_ylabel('Tarsal PER',fontsize=10)
ax.set_xlabel('[sucrose] (mM)',fontsize=10)
ax.set_ylim([0,0.8])
ax.set_xticks(list(range(len(data_suc.columns))),[x[:-2] for x in data_suc_geo.columns],fontsize=10)
```
