```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

fontsize = 7
lp = 2

concentrations = [10, 50, 100]
modalities = ['tarsal', 'labial']

cols_dict = {
    'tarsal': [
        'control',
        'roundup',
        'FDG',
        'sink and sync',
        'FMIn',
        'meteor',
    ],

    'labial': [
        'control',
        'FMIn',
        'roundup',
        'TH-VUM',
        'usnea',
        'clavicle',
        'bract',
        'rattle',
        'G2N-1',
        'meteor',
    ],
}

groups = [
    'suc_non_fixed',
    'suc_fixed',
]

ylabel_map = {
    'tarsal': 'Tarsal PER',
    'labial': 'Labellar PER',
}


# =====================================
# load data
# =====================================

all_data = {}

for modality in modalities:

    all_data[modality] = {}

    for conc in concentrations:

        fpath = f'data_fixed/{modality}_{conc}mM_raw.xlsx'

        all_data[modality][conc] = {
            s: pd.read_excel(fpath, sheet_name=s)
            for s in pd.ExcelFile(fpath).sheet_names
        }


# =====================================
# 각각 single-panel figure로 plot
# =====================================

for modality in modalities:

    cols = cols_dict[modality]

    for conc in concentrations:
        if modality == 'tarsal':
            size = (3,1.2)
        else:
            size = (6,1.2)
        # -----------------------------
        # one figure = one panel
        # -----------------------------
        fig, ax = plt.subplots(
            1,
            1,
            figsize=size
        )

        data_dict = all_data[modality][conc]


        # =====================================
        # mean / SEM
        # =====================================

        mean_mat = pd.DataFrame(
            index=cols,
            columns=groups,
            dtype=float
        )

        std_mat = pd.DataFrame(
            index=cols,
            columns=groups,
            dtype=float
        )

        for condition in groups:

            mean_mat[condition] = (
                data_dict[condition]
                .mean(axis=0)
                .reindex(cols)
            )

            std_mat[condition] = (
                pd.Series(
                    std_err(data_dict[condition]),
                    index=data_dict[condition].columns
                )
                .reindex(cols)
            )


        # =====================================
        # helper functions
        # =====================================

        def get_mean(cond, group, mean_mat=mean_mat):
            return mean_mat.loc[cond, group]


        def get_err(cond, group, std_mat=std_mat):
            return std_mat.loc[cond, group]


        def get_values(cond, group, data_dict=data_dict):
            return (
                data_dict[group][cond]
                .dropna()
                .to_numpy()
            )


        # =====================================
        # draw
        # =====================================

        x_loc = draw_bar_with_points(
            ax=ax,
            conditions=cols,
            groups=groups,

            get_values=get_values,
            get_mean=get_mean,
            get_err=get_err,

            bar_width=0.075,
            bar_spacing=0.3,

            colors=[
                cell2color[x]
                for x in cols
            ],

            edgecolor='black',
            linewidth=0.5,
            capsize=2,

            point_facecolor='white',
            point_edgecolor='black',
            point_lw=0.3,
            point_size=10,
            point_alpha=0.9,
        )


        # =====================================
        # axis
        # =====================================

        xlabel = [
            x.split(' ')[-1]
            for x in cols
        ]

        xlabel[0] = 'control'

        ax.set_xticks(
            x_loc,
            xlabel,
            fontsize=fontsize,
            rotation=45,
            ha='right',
        )

        yticks = np.arange(0, 1.0, 0.2)

        ax.set_yticks(
            yticks,
            np.round(yticks, 2),
            fontsize=fontsize,
        )

        ax.set_ylabel(
            ylabel_map[modality],
            fontsize=fontsize,
            labelpad=lp,
        )

        ax.set_ylim(
            0,
            0.85
        )

        ax.set_title(
            f'{conc} mM',
            fontsize=fontsize,
            pad=3,
        )

        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)

        ax.spines['left'].set_position(
            ('outward', 5)
        )

        if ax.get_legend() is not None:
            ax.get_legend().remove()


        plt.show()

        # 저장할 경우
        # plt.savefig(
        #     f'figures/{modality}_{conc}mM_fixed_compare.svg',
        #     bbox_inches='tight'
        # )
```
