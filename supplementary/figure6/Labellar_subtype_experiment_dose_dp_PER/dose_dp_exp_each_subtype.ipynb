```python
# receptor screening 

fpath = f'{os.getcwd()}/data/labellar_PER_each_subtype_stimulation_final.xlsx'
sheets = pd.ExcelFile(fpath)
sheets = sheets.sheet_names
data_df_receptor = {}
for s in sheets:
    df = pd.read_excel(fpath,s)
    data_df_receptor[s] = df
    
fpath = f'{os.getcwd()}/data/labellar_PER_each_subtype_stimulation_geosmin_final.xlsx'
sheets = pd.ExcelFile(fpath)
sheets = sheets.sheet_names
data_df_receptor_geosmin = {}
for s in sheets:
    df = pd.read_excel(fpath,s)
    data_df_receptor_geosmin[s] = df
```

```python
fig_params['ms'] = 5
fig_params['elw'] = 1
```

```python
# fig1c
fig,ax = plt.subplots(1,1,figsize=(1.26,1.4))
conditions = ['control','Ir56b_Kir2.1(L1+L2)']
alpha = [.5,1,.5]
linestyle = ['-.','-',':']
shape =  ['o','v','s']
for i,condition in enumerate(conditions):
    plot_dose_dp_PER(ax,data_df_receptor[condition],cidx=3,alpha=alpha[i],linestyle=linestyle[i],shape=shape[i],**fig_params)
    plot_dose_dp_PER(ax,data_df_receptor_geosmin[condition],cidx=2,alpha=alpha[i],linestyle=linestyle[i],shape=shape[i],**fig_params)


ax.set_ylabel('Labellar PER',fontsize=7)
ax.set_xlabel('[sucrose] (mM)',fontsize=7)
ax.set_ylim([0,0.8])
ax.set_xticks(list(range(len(data_df_receptor['control'].columns))),[x[:-2] for x in data_df_receptor['control'].columns],fontsize=7)
ax.tick_params(
    axis='both',
    width=.7,      # tick line 두께
)
for spine in ax.spines.values():
    spine.set_linewidth(.7)

# plt.savefig('L1L2.svg',bbox_inches='tight')
```
